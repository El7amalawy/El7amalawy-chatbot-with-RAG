## 🔧 Section 0: Setup - API Key Helper (Works Everywhere!)

This helper function works in:
- ✅ Google Colab (using Secrets)
- ✅ Local Python / Jupyter
- ✅ VS Code
- ✅ Any environment!

**Run this cell FIRST before anything else!**

````python
# ╔══════════════════════════════════════════════════════════════════════════════╗
# ║                         API Key Helper Function                              ║
# ╠══════════════════════════════════════════════════════════════════════════════╣
# ║                                                                              ║
# ║ إيه الـ Function دي؟                                                         ║
# ║   Function بتجيب الـ API Key من أي مكان تلقائياً                              ║
# ║                                                                              ║
# ║ بتدور على الـ Key في:                                                        ║
# ║   1. Google Colab Secrets (لو شغال في Colab)                                 ║
# ║   2. Environment Variables (لو محطوط في النظام)                              ║
# ║   3. .env file (لو موجود في المجلد)                                          ║
# ║   4. Manual Input (بتسألك تكتبه)                                             ║
# ║                                                                              ║
# ║ ليه عملناها كده؟                                                             ║
# ║   عشان الكود يشتغل في أي مكان من غير تعديل!                                  ║
# ║   - في Colab ✓                                                               ║
# ║   - في Jupyter local ✓                                                       ║
# ║   - في VS Code ✓                                                             ║
# ║                                                                              ║
# ╚══════════════════════════════════════════════════════════════════════════════╝

import os
from getpass import getpass

# Dictionary لتخزين الـ keys اللي جبناها
# عشان لو طلبنا نفس الـ key تاني منسألش المستخدم
_API_KEYS = {}

def get_api_key(key_name: str, display_name: str = None) -> str:
    """
    بتجيب الـ API Key من أي مكان.
    
    Parameters:
    -----------
    key_name : str
        اسم الـ key في النظام (مثلاً 'GOOGLE_API_KEY')
    
    display_name : str, optional
        الاسم اللي يظهر للمستخدم (مثلاً 'Gemini')
    
    Returns:
    --------
    str
        الـ API Key
    """
    
    # لو مفيش display_name، استخدم key_name
    if display_name is None:
        display_name = key_name
    
    # لو الـ key موجود في الـ cache، رجعه
    if key_name in _API_KEYS and _API_KEYS[key_name]:
        return _API_KEYS[key_name]
    
    api_key = None
    source = None
    
    # ────────────────────────────────────────────
    # المحاولة 1: Colab Secrets
    # ────────────────────────────────────────────
    # في Colab، تقدر تحط secrets في الـ sidebar
    # وتجيبها بـ userdata.get()
    try:
        from google.colab import userdata
        api_key = userdata.get(key_name)
        if api_key:
            source = "Colab Secrets"
    except:
        pass  # مش في Colab
    
    # ────────────────────────────────────────────
    # المحاولة 2: Environment Variable
    # ────────────────────────────────────────────
    # لو الـ key محطوط في environment variables
    # زي: export GOOGLE_API_KEY=xxx
    if not api_key:
        api_key = os.environ.get(key_name)
        if api_key:
            source = "Environment Variable"
    
    # ────────────────────────────────────────────
    # المحاولة 3: .env file
    # ────────────────────────────────────────────
    # لو فيه ملف .env في المجلد
    # فيه: GOOGLE_API_KEY=xxx
    if not api_key:
        try:
            from dotenv import load_dotenv
            load_dotenv()  # بتقرأ الـ .env file
            api_key = os.environ.get(key_name)
            if api_key:
                source = ".env file"
        except ImportError:
            pass  # مكتبة python-dotenv مش مثبتة
    
    # ────────────────────────────────────────────
    # المحاولة 4: اسأل المستخدم
    # ────────────────────────────────────────────
    if not api_key:
        print(f"🔑 {display_name} API Key مش موجود.")
        print(f"   جيب الـ key من: https://aistudio.google.com/apikey")
        api_key = getpass(f"   اكتب الـ {display_name} API Key: ")
        source = "Manual Input"
    
    # احفظ الـ key في الـ cache
    _API_KEYS[key_name] = api_key
    
    if source:
        print(f"✅ {display_name} API Key loaded from {source}")
    
    return api_key

print("✅ API Key Helper ready!")
````

````python
# 6. Error Handling (try/except = try/catch)
def safe_divide(a: float, b: float) -> float:
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return 0
    finally:
        print("Operation completed")

print(safe_divide(10, 2))   # 5.0
print(safe_divide(10, 0))   # Error message, returns 0
````

## 📚 Section 2: Chat UI vs API - Understanding the Difference

### 🎯 Why Use APIs Instead of Chat UIs?

| Aspect | Chat UI (ChatGPT, Gemini) | API |
|--------|---------------------------|-----|
| **Control** | Limited | Full control |
| **Integration** | Manual copy/paste | Automated workflows |
| **Customization** | Preset options | System prompts, parameters |
| **Scale** | One at a time | Batch processing |
| **Cost** | Subscription ($20/mo) | Pay per token (often cheaper) |
| **Privacy** | Data on provider servers | Can use local models |

---

### 🔑 API Key Concepts

```
┌─────────────────────────────────────────────────────────────┐
│                    API Request Flow                         │
├─────────────────────────────────────────────────────────────┤
│                                                             │
│   Your Code          API Endpoint           LLM Model       │
│   ─────────          ────────────           ─────────       │
│                                                             │
│   ┌─────────┐        ┌─────────┐          ┌─────────┐      │
│   │ Python  │───────►│ Google  │─────────►│ Gemini  │      │
│   │ Script  │ API Key│   API   │ Process  │  Model  │      │
│   └─────────┘◄───────└─────────┘◄─────────└─────────┘      │
│              Response                                       │
│                                                             │
└─────────────────────────────────────────────────────────────┘
```

### 📊 Token Economics

- **Token** ≈ 4 characters or ¾ of a word
- **Input tokens**: What you send (prompt + context)
- **Output tokens**: What the model generates
- **Pricing**: Usually per 1 million tokens

**Example:**
- "Hello, how are you?" ≈ 6 tokens
- A 500-word response ≈ 375 tokens

````python
# ============================================
# GROQ - Simple Chat Completion
# ============================================

import time

start_time = time.time()

chat_completion = groq_client.chat.completions.create(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant specialized in construction and BIM."
        },
        {
            "role": "user",
            "content": "What is Level of Development (LOD) in BIM? Explain briefly."
        }
    ],
    model="llama-3.3-70b-versatile",
    temperature=0.7,
    max_tokens=500,
)

elapsed = time.time() - start_time

print(chat_completion.choices[0].message.content)
print(f"\n⚡ Response time: {elapsed:.2f} seconds")
print(f"📊 Tokens used: {chat_completion.usage.total_tokens}")
````

````python
# ============================================
# GROQ - Streaming Response
# ============================================

stream = groq_client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": "List 5 benefits of using Python for BIM automation."
        }
    ],
    model="llama-3.1-8b-instant",  # Fastest model!
    temperature=0.5,
    stream=True,
)

print("Streaming response:")
for chunk in stream:
    content = chunk.choices[0].delta.content
    if content:
        print(content, end="", flush=True)

print("\n\n✅ Stream complete!")
````

````python
# ============================================
# GROQ - Multi-turn Conversation
# ============================================

# Build conversation history manually
messages = [
    {"role": "system", "content": "You are a Revit API expert."},
    {"role": "user", "content": "How do I get all walls in a Revit document?"}
]

# First response
response1 = groq_client.chat.completions.create(
    messages=messages,
    model="llama-3.3-70b-versatile"
)

assistant_msg = response1.choices[0].message.content
print("Bot:", assistant_msg[:500] + "..." if len(assistant_msg) > 500 else assistant_msg)

# Add to history
messages.append({"role": "assistant", "content": assistant_msg})
messages.append({"role": "user", "content": "Now filter only exterior walls."})

# Follow-up
response2 = groq_client.chat.completions.create(
    messages=messages,
    model="llama-3.3-70b-versatile"
)

print("\n" + "="*50 + "\n")
print("Bot:", response2.choices[0].message.content)
````

````python
# ============================================
# OPENROUTER - Compare Multiple Free Models
# ============================================

def ask_model(model_id: str, question: str) -> str:
    """Ask a question to any OpenRouter model."""
    try:
        response = openrouter_client.chat.completions.create(
            model=model_id,
            messages=[{"role": "user", "content": question}],
            max_tokens=200,
        )
        return response.choices[0].message.content
    except Exception as e:
        return f"Error: {str(e)}"

# Test with different FREE models
question = "What is IFC in one sentence?"

models = [
    "meta-llama/llama-3.3-70b-instruct:free",
    "qwen/qwen-2.5-72b-instruct:free",
    "deepseek/deepseek-chat:free",
]

print(f"Question: {question}\n")
print("="*50)

for model in models:
    model_name = model.split('/')[1].split(':')[0]
    print(f"\n🤖 {model_name}:")
    print(ask_model(model, question))
    print("-"*50)
````

| `st.write_stream()` | Streaming output |

````python
# ============================================
# STREAMLIT CHATBOT - app.py
# ============================================

# Save this to a file called 'app.py' and run with: streamlit run app.py

streamlit_code = '''import streamlit as st
from groq import Groq
import os

# ============================================
# PAGE CONFIG
# ============================================
st.set_page_config(
    page_title="ITI ChatBot",
    page_icon="🤖",
    layout="centered"
)

st.title("🤖 ITI ChatBot")
st.caption("Powered by Groq + Llama 3.3")

# ============================================
# SIDEBAR - Settings
# ============================================
with st.sidebar:
    st.header("⚙️ Settings")
    
    # API Key input (checks env var first)
    default_key = os.environ.get("GROQ_API_KEY", "")
    api_key = st.text_input(
        "Groq API Key",
        type="password",
        value=default_key,
        help="Get your key from console.groq.com"
    )
    
    # Model selection
    model = st.selectbox(
        "Model",
        ["llama-3.3-70b-versatile", "llama-3.1-8b-instant", "llama-4-scout-17b"]
    )
    
    # Temperature slider
    temperature = st.slider(
        "Temperature",
        min_value=0.0,
        max_value=2.0,
        value=0.7,
        step=0.1
    )
    
    # System prompt
    system_prompt = st.text_area(
        "System Prompt",
        value="You are a helpful assistant specialized in BIM and construction.",
        height=100
    )
    
    # Clear chat button
    if st.button("🗑️ Clear Chat"):
        st.session_state.messages = []
        st.rerun()

# ============================================
# INITIALIZE SESSION STATE
# ============================================
if "messages" not in st.session_state:
    st.session_state.messages = []

# ============================================
# DISPLAY CHAT HISTORY
# ============================================
for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

# ============================================
# CHAT INPUT
# ============================================
if prompt := st.chat_input("Type your message..."):
    # Check API key
    if not api_key:
        st.error("Please enter your Groq API key in the sidebar!")
        st.stop()
    
    # Add user message to history
    st.session_state.messages.append({"role": "user", "content": prompt})
    
    # Display user message
    with st.chat_message("user"):
        st.markdown(prompt)
    
    # Generate response
    with st.chat_message("assistant"):
        try:
            # Initialize Groq client
            client = Groq(api_key=api_key)
            
            # Build messages list
            messages = [{"role": "system", "content": system_prompt}]
            messages.extend(st.session_state.messages)
            
            # Stream response
            stream = client.chat.completions.create(
                model=model,
                messages=messages,
                temperature=temperature,
                stream=True,
            )
            
            # Display streaming response
            response = st.write_stream(
                (chunk.choices[0].delta.content or "" for chunk in stream)
            )
            
            # Add assistant response to history
            st.session_state.messages.append({"role": "assistant", "content": response})
            
        except Exception as e:
            st.error(f"Error: {str(e)}")
'''

# Save to file
with open('app.py', 'w') as f:
    f.write(streamlit_code)

print("✅ app.py created!")
print("\n" + "="*50)
print("To run locally:")
print("  1. pip install streamlit groq")
print("  2. streamlit run app.py")
print("="*50)
````

### 📊 Token Counting

````python
# ============================================
# TOKEN COUNTING WITH TIKTOKEN
# ============================================

!pip install -q tiktoken

import tiktoken

def count_tokens(text: str, model: str = "gpt-4") -> int:
    """Count tokens in text using tiktoken."""
    try:
        encoding = tiktoken.encoding_for_model(model)
    except:
        encoding = tiktoken.get_encoding("cl100k_base")  # Default
    
    return len(encoding.encode(text))

# Test
sample_text = "Building Information Modeling (BIM) is a digital representation of physical and functional characteristics of a facility."

tokens = count_tokens(sample_text)
print(f"Text: {sample_text}")
print(f"\nTokens: {tokens}")
print(f"Characters: {len(sample_text)}")
print(f"Ratio: {len(sample_text)/tokens:.1f} chars per token")
````

### 🔄 Error Handling & Retries

````python
# ============================================
# ROBUST API CALLS WITH RETRIES
# ============================================

import time
from typing import Callable, Any

def retry_with_backoff(
    func: Callable,
    max_retries: int = 3,
    base_delay: float = 1.0,
    max_delay: float = 60.0
) -> Any:
    """Retry a function with exponential backoff."""
    
    for attempt in range(max_retries):
        try:
            return func()
        except Exception as e:
            if attempt == max_retries - 1:
                raise e
            
            # Calculate delay with exponential backoff
            delay = min(base_delay * (2 ** attempt), max_delay)
            print(f"Attempt {attempt + 1} failed: {e}")
            print(f"Retrying in {delay:.1f} seconds...")
            time.sleep(delay)

# Example usage
def make_api_call():
    # Your API call here
    return "Success!"

result = retry_with_backoff(make_api_call)
print(result)
````

## 📚 Section 9: Best Practices

### ✅ Do's

1. **Store API keys in environment variables** - Never hardcode!
2. **Use streaming** for better user experience
3. **Implement retries** with exponential backoff
4. **Monitor token usage** to control costs
5. **Cache responses** when appropriate
6. **Set reasonable timeouts** (30-60 seconds)
7. **Handle rate limits gracefully**

### ❌ Don'ts

1. **Don't commit API keys** to Git repositories
2. **Don't ignore errors** - always handle exceptions
3. **Don't make unlimited retries** - set a max
4. **Don't trust user input** - validate and sanitize
5. **Don't forget to set max_tokens** - prevent runaway costs
